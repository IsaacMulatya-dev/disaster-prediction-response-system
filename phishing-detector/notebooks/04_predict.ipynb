```python
import sys
sys.path.append("../src")

from predict import load_model_and_vectorizer, predict_batch
from data_preprocessing import clean_email_text
import pandas as pd

# Load model and vectorizer
model, vectorizer = load_model_and_vectorizer()

# Sample list of raw emails
emails = [
    "Your account has been compromised. Click here to reset your password.",
    "Hi team, please find the meeting notes attached.",
    "Verify your identity immediately at this link: http://scam-link.com",
    "Reminder: Project kickoff meeting tomorrow at 10AM."
]

# Clean each email
cleaned_emails = [clean_email_text(email) for email in emails]

# Predict in batch
predictions = predict_batch(cleaned_emails, model, vectorizer)

# Show results in a table
results_df = pd.DataFrame({
    "Original Email": emails,
    "Cleaned Email": cleaned_emails,
    "Prediction": ["Phishing" if p == 1 else "Legitimate" for p in predictions]
})

results_df
```
